```python
import numpy as np
import pandas as pd

# Create a random NumPy array
np.random.seed(42)  # For reproducibility
data = np.random.randint(1, 101, size=(10, 4))  # 10 rows, 4 columns

# Convert the NumPy array to a Pandas DataFrame
columns = ['A', 'B', 'C', 'D']
df = pd.DataFrame(data, columns=columns)

# Add a new column with the mean of each row (using NumPy)
df['Row_Mean'] = np.mean(data, axis=1)

# Add another column with the square root of column 'A' (using NumPy)
df['A_Sqrt'] = np.sqrt(df['A'])

# Filter rows where the mean is greater than 50
filtered_df = df[df['Row_Mean'] > 50]

# Display the original and filtered DataFrames
print("Original DataFrame:")
print(df)
print("\nFiltered DataFrame (Row_Mean > 50):")
print(filtered_df)
```
